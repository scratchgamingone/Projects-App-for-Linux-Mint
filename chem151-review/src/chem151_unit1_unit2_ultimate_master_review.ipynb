## 1. Complete Dictionary of Definitions (Unit 1 & Unit 2)

### Unit 1: Matter, Measurement, Atoms, Chemical Reactions & Stoichiometry
- **Chemistry:** The study of matter and the physical and chemical changes that it undergoes.
- **Matter:** Anything that has mass and occupies space (volume).
- **States of Matter:**
  - **Solid:** Definite shape and definite volume; particles are tightly packed in rigid, fixed positions.
  - **Liquid:** Definite volume but no fixed shape (assumes the shape of its container); particles are in close contact but free to move relative to one another.
  - **Gas:** No fixed volume or shape; particles are widely separated and in complete freedom of motion; easily compressed and expanded.
- **The Scientific Method:**
  - **Observation:** Qualitative or quantitative data collected about natural phenomena or experiments.
  - **Hypothesis:** A *tentative* explanation for a set of observations.
  - **Law:** A concise mathematical or verbal *summary* of a broad set of observations (e.g., Law of Conservation of Mass).
  - **Theory:** A comprehensively *tested and verified explanation* for observed phenomena.
- **Accuracy vs. Precision:**
  - **Accuracy:** How close a measured value is to the true or accepted value (hitting the bullseye).
  - **Precision:** How close individual repeated measurements are to each other (reproducibility).
- **Mass vs. Weight:**
  - **Mass:** Invariant measure of the quantity of matter in an object; measured using a balance.
  - **Weight:** Force exerted on an object by a gravitational field; varies by location; measured using a scale.
- **Matter Classification Scheme:**
  - **Pure Substance:** Matter possessing a constant composition and uniform properties; cannot be separated by physical means.
  - **Element:** Substance composed of one kind of atom that cannot be decomposed into simpler substances by chemical reactions.
  - **Compound:** Substance composed of two or more distinct elements chemically combined in fixed ratios; can only be decomposed by chemical reactions.
  - **Mixture:** Physical blend of two or more pure substances in variable proportions, retaining their individual identities; separable by physical methods.
  - **Homogeneous Mixture (Solution):** Mixture that is uniform in composition throughout to the naked eye (e.g., salt water, air, brass).
  - **Heterogeneous Mixture:** Mixture lacking uniform composition, possessing visibly distinct phases (e.g., sand in water, trail mix, oil and water).
- **Physical vs. Chemical Change:**
  - **Physical Change:** Change in state or appearance without any change in chemical composition (reversible; e.g., melting ice, boiling water, dissolving sugar).
  - **Chemical Change (Reaction):** Change where substances convert into new substances with different compositions and properties (irreversible by physical means; e.g., combustion, rusting iron, milk souring).
  - *Clues of chemical reactions:* Color change, gas bubbles, temperature change, precipitate formation.
- **Significant Figures:** The reliably known digits in a measured quantity plus one final estimated/uncertain digit.
- **Exact Numbers:** Numbers known with complete certainty bearing infinite significant figures (e.g., counted items, definitions like 1 in = 2.54 cm).
- **Density ($D$):** Intensive physical property defined as the mass of a substance per unit volume ($D = M/V$).
- **Atom:** The fundamental chemical building block of matter consisting of a dense central nucleus surrounded by electrons.
- **Subatomic Particles:**
  - **Proton ($p$):** Positively charged subatomic particle residing in the nucleus ($+1.60 \times 10^{-19}\text{ C}$, mass $\approx 1.673 \times 10^{-24}\text{ g} \approx 1.007\text{ amu}$).
  - **Neutron ($n$):** Electrically neutral subatomic particle residing in the nucleus (mass $\approx 1.675 \times 10^{-24}\text{ g} \approx 1.008\text{ amu}$).
  - **Electron ($e^-$):** Negatively charged particle outside the nucleus ($-1.60 \times 10^{-19}\text{ C}$, mass $\approx 9.11 \times 10^{-28}\text{ g} \approx 5.486 \times 10^{-4}\text{ amu}$, negligible mass).
- **Atomic Number ($Z$):** Total number of protons in the nucleus of an atom; defines the chemical identity of the element.
- **Mass Number ($A$):** Total number of nucleons (protons $+$ neutrons) in the nucleus of a specific atom.
- **Isotopes:** Atoms of the identical element sharing the same atomic number ($Z$) but differing in mass number ($A$) due to different numbers of neutrons.
- **Average Atomic Mass / Atomic Weight:** The weighted average mass of naturally occurring isotopes of an element, based on percent natural abundance.
- **Ion:** An electrically charged chemical species formed when a neutral atom gains or loses electrons.
  - **Cation:** Positively charged ion formed by the loss of electrons (metals on the left side of the periodic table).
  - **Anion:** Negatively charged ion formed by the gain of electrons (nonmetals on the right side of the periodic table).
- **Polyatomic Ion:** A tightly bound group of nonmetal atoms possessing an overall net ionic charge.
- **Diatomic Molecule:** A molecule composed naturally of exactly two atoms ($H_2, N_2, O_2, F_2, Cl_2, Br_2, I_2$).
- **The Mole:** The SI base unit for amount of substance containing exactly $6.022 \times 10^{23}$ representative entities.
- **Molar Mass:** The mass in grams of one mole of a chemical substance (g/mol), numerically equal to formula mass in amu.
- **Empirical Formula:** The simplest whole-number ratio of atoms of each element present in a compound.
- **Molecular Formula:** The actual whole-number formula representing the exact count of each atom in a molecule.
- **Percent Composition:** The percentage by mass contributed by each element to the total mass of a compound.
- **Combination Reaction:** Two or more reactants combine to yield a single compound ($A + B \to AB$).
- **Decomposition Reaction:** A single reactant breaks down into two or more simpler substances ($AB \to A + B$).
- **Combustion Reaction:** Rapid exothermic oxidation of a hydrocarbon/organic compound with $O_2(g)$ yielding $CO_2(g)$ and $H_2O(g)$.
- **Limiting Reactant (LR):** The reactant completely consumed first in a reaction, thereby limiting the maximum amount of product formed.
- **Excess Reactant:** The reactant remaining unconsumed once the limiting reactant has completely reacted.
- **Theoretical Yield:** The maximum quantity of product calculated to form through stoichiometric conversion of the limiting reactant.
- **Actual Yield:** The actual quantity of product isolated and weighed in a laboratory experiment.
- **Percent Yield:** The ratio of actual yield to theoretical yield expressed as a percentage ($[\text{Actual} / \text{Theoretical}] \times 100\%$).

### Unit 2: Electronic Structure, Periodic Trends, Bonding, Geometry & MOT
- **Electromagnetic Radiation:** Form of radiant energy propagating through a vacuum at the speed of light ($2.9979 \times 10^8\text{ m/s}$) exhibiting wave-like periodicity.
- **Wavelength ($\lambda$):** The linear distance between two consecutive crests or troughs of an electromagnetic wave (m or nm).
- **Frequency ($\nu$):** The number of complete wave cycles that pass a given point each second ($s^{-1}$ or Hz).
- **Amplitude:** The vertical height from node to crest; dictates radiant intensity/brightness.
- **Quantum:** The minimum discrete packet of energy that can be gained or lost by an atom.
- **Photon:** A quantum of electromagnetic energy behaving as a particle with energy proportional to frequency ($E = h\nu$).
- **Photoelectric Effect:** The emission of electrons from a clean metal surface when illuminated by light possessing sufficient threshold frequency.
- **Continuous Spectrum:** A spectrum displaying an unbroken continuum of all wavelengths (e.g., rainbow, white light from sun/lamp).
- **Line Spectrum:** A spectrum displaying sharp, discrete, isolated emission wavelengths unique to a specific gaseous element.
- **Ground State:** The lowest possible energy electronic arrangement of an atom ($n=1$ in Bohr hydrogen).
- **Excited State:** Any electronic state possessing higher energy than the ground state ($n > 1$).
- **Atomic Orbital:** A three-dimensional spatial probability distribution (wavefunction squared, $\psi^2$) where an electron has up to a 90% probability of being found.
- **Node:** A region or plane in an orbital where the probability of finding an electron is identically zero ($\psi^2 = 0$).
- **Quantum Numbers:** Set of four values ($n, l, m_l, m_s$) completely describing the energy, shape, orientation, and spin of an electron.
- **Degenerate Orbitals:** Two or more orbitals possessing identical energy levels (e.g., the three $2p$ orbitals in an isolated atom).
- **Core Electrons:** Inner-shell electrons occupying completely filled noble gas electron shells underneath the valence shell.
- **Valence Electrons:** Electrons occupying the outermost principal quantum shell; dictate chemical bonding and periodic behavior.
- **Bonding Atomic Radius:** One-half of the distance between the nuclei of two identical atoms covalently bonded together ($r = \frac{1}{2}d$).
- **Effective Nuclear Charge ($Z_{\text{eff}}$):** The net inward positive electrostatic attraction experienced by a valence electron from the nucleus ($Z_{\text{eff}} = Z - S$).
- **Electron Shielding (Screening):** The repulsive screening effect exerted by inner core electrons reducing the full nuclear charge felt by outer valence electrons.
- **First Ionization Energy ($I_1$):** Minimum energy required to remove the highest-energy valence electron from an isolated gaseous ground-state atom (always endothermic).
- **Successive Ionization Energies ($I_2, I_3, \dots$):** Energy required to remove subsequent electrons from positive ions ($I_1 < I_2 < I_3$).
- **Core Electron Quantum Leap:** The massive multi-fold surge in ionization energy observed when an element runs out of valence electrons and core electrons are removed.
- **Electron Affinity ($EA$):** The energy change associated with adding an electron to an isolated gaseous atom to form an anion.
- **Electronegativity ($EN$):** The relative empirical ability of an atom in a molecule to attract shared bonding electrons toward itself.
- **Metallic Character:** Physical and chemical tendencies associated with metals (luster, malleability, low ionization energy, cation formation, basic oxides).
- **Isoelectronic Series:** A family of atoms and ions possessing the exact same total number of electrons and identical ground-state electron configurations.
- **Ionic Bond:** Electrostatic attraction holding oppositely charged cations and anions together following electron transfer.
- **Covalent Bond:** Chemical bond formed through mutual sharing of one or more electron pairs between two nonmetal nuclei.
- **Metallic Bond:** Delocalized sharing of valence electrons within a mobile electron sea across a metal lattice.
- **Bond Length:** The equilibrium distance between the nuclei of two bonded atoms where electrostatic attraction balances repulsion.
- **Bond Enthalpy (Dissociation Energy):** The energy required to break one mole of a specific covalent bond in the gas phase (always positive/endothermic).
- **Octet Rule:** Tendency of main-group atoms to gain, lose, or share electrons until surrounded by 8 valence electrons ($ns^2 np^6$).
- **Bonding Pair:** A pair of shared valence electrons participating directly in a covalent bond.
- **Lone Pair (Nonbonding Pair):** An unshared pair of valence electrons localized on a single atom.
- **Formal Charge:** The theoretical charge assigned to an individual atom in a Lewis structure assuming equal sharing of shared bonds.
- **Resonance Structures:** Two or more valid Lewis dot structures representing the same molecule with identical atom connectivity but differing electron positions.
- **Resonance Hybrid:** The true, blended molecular electronic structure representing the weighted average of all contributing resonance forms.
- **Electron Domain:** A region of high electron density around a central atom (every single bond, double bond, triple bond, or lone pair = 1 domain).
- **Electron Pair Arrangement (Electron-Domain Geometry):** The 3D spatial distribution of all electron domains around the central atom.
- **Molecular Geometry (Shape):** The actual 3D arrangement of only the bonded atoms around the central atom.
- **Valence Shell Electron Pair Repulsion (VSEPR) Theory:** Model postulating that electron domains arrange around a central atom to minimize mutual electrostatic repulsions.
- **Hybridization:** The quantum mathematical mixing of atomic orbitals on a central atom to produce equivalent hybrid orbitals matching observed bond angles.
- **Sigma ($\sigma$) Bond:** A covalent bond formed by direct head-on overlap of orbitals along the internuclear axis.
- **Pi ($\pi$) Bond:** A covalent bond formed by parallel sideways overlap of unhybridized $p$-orbitals above and below the internuclear axis.
- **Dipole Moment ($\mu$):** Quantitative measure of the separation of positive and negative charges in a polar molecule ($\mu = Q \times r$).
- **Molecular Polarity:** Property of an asymmetrical molecule where individual bond dipoles fail to cancel, yielding a net molecular dipole moment.
- **Molecular Orbital Theory (MOT):** Model describing electrons in molecules by wavefunctions delocalized across the entire molecule rather than individual atoms.
- **Bonding Molecular Orbital:** Lower-energy molecular orbital formed by constructive wave interference that concentrates electron density between nuclei.
- **Antibonding Molecular Orbital ($\sigma^*, \pi^*$):** Higher-energy molecular orbital formed by destructive wave interference that introduces a node between nuclei.
- **Bond Order (MO):** Metric measuring bond stability and strength equal to one-half the difference between bonding and antibonding electrons.
- **Diamagnetism:** Property of substances with all electrons paired; weakly repelled by external magnetic fields.
- **Paramagnetism:** Property of substances with one or more unpaired electrons; attracted into external magnetic fields.

### Complete Historical Registry of Chemistry

| :--- | :--- | :--- | :--- |
| **Antoine Lavoisier** (1743–1794) | Quantitative closed-vessel combustion | Established the **Law of Conservation of Mass** | Total mass of products identically equals total mass of reactants ($m_{\text{reactants}} = m_{\text{products}}$). Disproved the phlogiston theory. |

| **Max Planck** (1900) | Blackbody radiation emissions | Proposed **Quantization of Energy**; discovered **Planck's Constant** ($h = 6.626 \times 10^{-34}\text{ J}\cdot\text{s}$) | Energy is emitted/absorbed only in discrete whole-number multiples of quanta ($E = nh\nu$). |
| **Albert Einstein** (1905) | **Photoelectric Effect** analysis | Introduced the **Photon**; established wave-particle duality of light ($E = h\nu = hc/\lambda$) | Light behaves as a localized stream of energy packets. One photon dislodges one electron if frequency surpasses threshold. |
| **Niels Bohr** (1913) | Hydrogen atomic line spectrum | Proposed the **Bohr Model of the Atom**; quantized angular momentum orbits | Electrons occupy stationary circular orbits with discrete energy levels ($E_n = -R_H/n^2$). Explains hydrogen line spectra, but fails for $Z \ge 2$. |

| **Wolfgang Pauli** (1925) | Atomic spectroscopic analysis | Established the **Pauli Exclusion Principle** | No two electrons in an atom can have the identical set of 4 quantum numbers ($n, l, m_l, m_s$). An orbital holds at most 2 electrons with opposite spins. |
| **Friedrich Hund** (1927) | Spectral term analysis | Established **Hund's Rule of Maximum Multiplicity** | In degenerate orbitals ($p, d, f$), electrons occupy orbitals singly with parallel spins before pairing to minimize repulsions. |
| **Gilbert N. Lewis** (1916) | Chemical bonding electronic theory | Introduced **Lewis Dot Symbols**, covalent electron pair sharing, and the **Octet Rule** | Valence electrons govern bonding; atoms gain, lose, or share electrons to reach a stable noble gas configuration ($ns^2 np^6$). |

### Guiding Principles & Electronic Ground Rules
1. **Aufbau Principle:** Orbitals fill strictly in order of increasing energy level: $1s < 2s < 2p < 3s < 3p < 4s < 3d < 4p < 5s < 4d < 5p < 6s < 4f \dots$
2. **Pauli Exclusion Principle:** Each atomic orbital holds a maximum of two electrons, and they must have opposite spins ($\uparrow\downarrow$).
3. **Hund's Rule:** In orbitals of equal energy (degenerate subshells), electrons must occupy each orbital singly with parallel spins before pairing.
4. **Stability of Half-Filled / Filled $d$-Subshells:** When an atom is one electron away from a half-filled ($d^5$) or completely filled ($d^{10}$) subshell, an $s$-electron promotes into the $d$-subshell:
   - **Chromium (Cr, $Z=24$):** $[\text{Ar}]\,4s^1 3d^5$ (not $4s^2 3d^4$)
   - **Copper (Cu, $Z=29$):** $[\text{Ar}]\,4s^1 3d^{10}$ (not $4s^2 3d^9$)
   - **Molybdenum (Mo, $Z=42$):** $[\text{Kr}]\,5s^1 4d^5$
   - **Silver (Ag, $Z=47$):** $[\text{Kr}]\,5s^1 4d^{10}$
5. **Cation Removal Priority Rule:** Transition and post-transition metals lose electrons from the outermost valence shell (highest $n$, the $s$ and $p$ electrons) **first** before losing any $(n-1)d$ electrons:
   - $\text{Sn} \to \mathbf{Sn^{2+}: [\text{Kr}]\,5s^2 4d^{10}}$ (loses $5p^2$)
   - $\text{Ni} \to \mathbf{Ni^{2+}: [\text{Ar}]\,3d^8}$ (loses $4s^2$)
   - $\text{Fe} \to \mathbf{Fe^{2+}: [\text{Ar}]\,3d^6}$ and $\mathbf{Fe^{3+}: [\text{Ar}]\,3d^5}$ (loses $4s^2$ then one $3d$)

## 4. Master Formulas, Variables, Calculations & Periodic Trends

### 1. Unit 1 Formulas & Equations
- **Density Formula:**
  $$D = \frac{M}{V} \implies M = D \times V \implies V = \frac{M}{D}$$
- **Average Atomic Mass / Atomic Weight:**
  $$\text{Average Atomic Mass} = \sum (\text{fractional abundance}) \times (\text{mass of isotope in amu})$$
- **Mass Percent Composition:**
  $$\text{Mass \% of element} = \frac{(\text{number of atoms}) \times (\text{molar mass of element})}{\text{molar mass of compound}} \times 100\%$$
- **Empirical to Molecular Formula Multiple ($n$):**
  $$n = \frac{\text{Molar mass of compound}}{\text{Empirical formula mass}} \implies \text{Molecular Formula} = n \times (\text{Empirical Formula})$$
- **Percent Yield:**
  $$\text{Percent Yield} = \frac{\text{Actual Yield}}{\text{Theoretical Yield}} \times 100\%$$

---

### 2. Unit 2 Formulas & Equations
- **Electromagnetic Wave Relationship:**
  $$c = \lambda \nu \implies \nu = \frac{c}{\lambda} \implies \lambda = \frac{c}{\nu}$$
- **Planck-Einstein Photon Energy:**
  $$E = h\nu = \frac{hc}{\lambda}$$
- **Bohr Electronic State Energy (Hydrogen):**
  $$E_n = \frac{-2.178 \times 10^{-18} \text{ J}}{n^2}$$
- **Bohr Electronic Transition Energy:**
  $$\Delta E = E_f - E_i = -2.178 \times 10^{-18} \text{ J} \left( \frac{1}{n_f^2} - \frac{1}{n_i^2} \right)$$
  - $\Delta E < 0$: Photon **emitted** (electron drops down, $n_i > n_f$)
  - $\Delta E > 0$: Photon **absorbed** (electron promoted, $n_i < n_f$)
  - Emitted/absorbed wavelength: $\lambda = \frac{hc}{|\Delta E|}$
- **de Broglie Matter Wavelength:**
  $$\lambda = \frac{h}{mv} \quad (m \text{ in kg}, v \text{ in m/s})$$
- **Shell Capacity Quantum Rules:**
  - Total orbitals in shell $n = n^2$
  - Maximum electrons in shell $n = 2n^2$
- **Effective Nuclear Charge:**
  $$Z_{\text{eff}} = Z - S \quad (Z = \text{protons}, S = \text{core shielding electrons})$$
- **Covalent Atomic Radius:**
  $$r = \frac{1}{2}d \quad (d = \text{internuclear bond distance})$$
- **Formal Charge:**
  $$\text{Formal Charge} = (\text{Valence } e^- \text{ on Periodic Table}) - (\text{Lone pair } e^-) - (\text{Number of bonds})$$
- **Enthalpy of Reaction from Bond Enthalpies:**
  $$\Delta H_{\text{rxn}} = \sum \text{BE}(\text{broken bonds in reactants}) - \sum \text{BE}(\text{formed bonds in products})$$
- **Enthalpy of Reaction from Standard Heats of Formation:**
  $$\Delta H^\circ_{\text{rxn}} = \sum n\Delta H_f^\circ(\text{products}) - \sum m\Delta H_f^\circ(\text{reactants})$$
- **Molecular Orbital Bond Order:**
  $$\text{Bond Order} = \frac{(\text{Bonding } e^-) - (\text{Antibonding } e^-)}{2}$$

### 3. The Master Periodic Table Vector Map & Shortcut Methods

```
        UP & RIGHT (↗ toward F / He)             DOWN & LEFT (↙ toward Fr / Cs)
    ------------------------------------       ----------------------------------
    • Ionization Energy increases              • Atomic Radius increases
    • Electron Affinity increases              • Metallic Character increases
    • Electronegativity increases
```

#### The 2-Step Inspection Method for Any Element Comparison
1. **Check Period ($n$) First (Vertical Test):**
   - Elements lower down have higher shell numbers ($n$).
   - Lower down = **Larger atomic radius** and **Lower ionization energy / electronegativity**.
2. **Check Group ($Z_{\text{eff}}$) Second (Horizontal Test):**
   - Across a row, period is constant; protons increase to the right.
   - Farther right = **Higher $Z_{\text{eff}}$**, pulling electrons tighter $\implies$ **Smaller atomic radius**, **Higher ionization energy**, and **Higher electronegativity**.

#### Isoelectronic Series Ranking (Smallest to Largest Radius)
- **Rule: MORE PROTONS = SMALLER RADIUS**
- When species have identical electron counts (e.g., 10 electrons):
  $$\mathbf{Al^{3+}} (13p) < \mathbf{Mg^{2+}} (12p) < \mathbf{Na^+} (11p) < \mathbf{F^-} (9p) < \mathbf{O^{2-}} (8p) < \mathbf{N^{3-}} (7p)$$

#### Successive Ionization Energies & Quantum Leap
- When all valence electrons are removed, breaking into the inner noble-gas core causes a dramatic multi-thousand $\text{kJ/mol}$ surge.
- Count the low IE values before the leap to determine the Group Number:
  - Jump between $I_1$ and $I_2 \implies 1$ valence $e^- \implies$ **Group 1A** (e.g., Na)
  - Jump between $I_2$ and $I_3 \implies 2$ valence $e^- \implies$ **Group 2A** (e.g., Mg)
  - Jump between $I_3$ and $I_4 \implies 3$ valence $e^- \implies$ **Group 3A** (e.g., Al)

## 5. Chemical Bonding, 8-Step Lewis Dot Structures, VSEPR Shapes, Hybridization & MOT

### The 8-Step Lewis Dot Structure Protocol (Chapter 8)
1. **Determine the Skeleton:**
   - Least electronegative atom is central (never Hydrogen).
   - Hydrogen is always terminal (forms only 1 single bond, max 2 electrons).
   - Halogens are generally terminal (unless bonded to O or a heavier halogen).
   - Symmetrical formulas have symmetrical skeletons.
   - In oxyacids, H bonds to O, and O bonds to the central atom (e.g., $\text{O—N(O)—O—H}$ for $\text{HNO}_3$).
2. **Count Total Valence Electrons:**
   - Sum valence electrons for all atoms using periodic table group numbers.
   - Adjust for ionic charges: Add electrons for negative charges; subtract for positive charges.
3. **Draw Single Bonds:**
   - Connect central atom to each terminal atom with single bonds ($2e^-$ per bond).
4. **Complete Octets on Terminal Atoms:**
   - Place up to 3 lone pairs ($6e^-$) on each terminal atom (except H).
5. **Place Leftover Electrons on the Central Atom:**
   - Subtract used electrons from total. Place any leftovers directly on the central atom as lone pairs.
6. **Satisfy Central Octet Using Multiple Bonds:**
   - If central atom has $< 8e^-$, convert lone pairs on terminal atoms into double or triple bonds (common for C, N, O, S).
7. **Account for Octet Exceptions:**
   - **Incomplete Octets:** Beryllium (Be) is stable with 4 electrons (2 bonds); Boron (B) is stable with 6 electrons (3 bonds, e.g., $\text{BF}_3$).
   - **Odd-Electron Molecules:** Radicals possessing odd valence electrons (e.g., $\text{NO}$ with $11e^-$).
   - **Expanded Octets:** Central atoms in **Period 3 and below** (P, S, Cl, Br, I, Xe) have accessible $d$-orbitals and can hold 10, 12, or 14 electrons (e.g., $\text{SF}_6, \text{XeF}_4$).
8. **Evaluate Formal Charges & Resonance:**
   $$\text{Formal Charge} = (\text{Valence } e^-) - (\text{Lone pair electrons}) - (\text{Number of bonds})$$
   - Structures with formal charges closest to 0 are most stable.
   - Negative formal charge should reside on the most electronegative atom.
   - Enclose polyatomic ions and resonance forms in square brackets with double-headed arrows ($\longleftrightarrow$).

- **Molecular Polarity:** Molecule is polar if bonds are polar AND shape is asymmetrical (terminal atoms differ OR central atom has uncancelled lone pairs).

### Molecular Orbital Theory (MOT) Summary
- **Bond Order:** $\frac{\text{Bonding } e^- - \text{Antibonding } e^-}{2}$
- **Bond Order Stability:** $0$ = unstable/no bond; $1$ = single bond; $2$ = double bond; $3$ = triple bond.
- **Diamagnetic:** All electrons paired (weakly repelled by magnetic field, e.g., $\text{N}_2$).
- **Paramagnetic:** $\ge 1$ unpaired electrons (attracted into magnetic field, e.g., $\text{O}_2$).

````python
# Tool 1: Comprehensive Unit & Temperature Converter
# Covers Metric Prefixes, English-Metric Equivalence, and Temperature Scales

METRIC_PREFIXES = {
    "G": ("giga", 1e9), "M": ("mega", 1e6), "k": ("kilo", 1e3),
    "d": ("deci", 1e-1), "c": ("centi", 1e-2), "m": ("milli", 1e-3),
    "u": ("micro", 1e-6), "n": ("nano", 1e-9), "p": ("pico", 1e-12), "": ("base", 1.0)
}

def convert_metric_prefix(value, from_pfx, to_pfx):
    """Converts a numeric value between metric prefixes (e.g. 'k' to 'm')."""
    base_val = value * METRIC_PREFIXES[from_pfx][1]
    return f"{base_val / METRIC_PREFIXES[to_pfx][1]:.4e}"

def convert_temp(val, from_scale, to_scale):
    """Converts temperature between Fahrenheit ('F'), Celsius ('C'), and Kelvin ('K')."""
    from_s, to_s = from_scale.upper(), to_scale.upper()
    c = (val - 32) / 1.8 if from_s == "F" else (val if from_s == "C" else val - 273.15)
    if to_s == "C": return round(c, 2)
    elif to_s == "F": return round(1.8 * c + 32, 2)
    elif to_s == "K": return round(c + 273.15, 2)

def convert_english_metric(val, unit_from, unit_to):
    """Common CHEM 151 conversion factors: in<->cm, lb<->g, gal<->L, atm<->mmHg/torr."""
    factors = {
        ("in", "cm"): 2.54, ("cm", "in"): 1 / 2.54,
        ("lb", "g"): 453.592, ("g", "lb"): 1 / 453.592,
        ("gal", "l"): 3.78541, ("l", "gal"): 1 / 3.78541,
        ("atm", "mmhg"): 760.0, ("mmhg", "atm"): 1 / 760.0,
        ("atm", "torr"): 760.0, ("torr", "atm"): 1 / 760.0,
        ("atm", "kpa"): 101.325, ("kpa", "atm"): 1 / 101.325
    }
    key = (unit_from.lower(), unit_to.lower())
    if key in factors:
        return round(val * factors[key], 4)
    raise ValueError(f"Conversion from {unit_from} to {unit_to} not supported.")

print("--- Tool 1: Unit & Temperature Conversions ---")
print("1 kg to mg:       ", convert_metric_prefix(1, "k", "m"), "mg")
print("1 ds to ps:       ", convert_metric_prefix(1, "d", "p"), "ps")
print("72 °F to Celsius: ", convert_temp(72, "F", "C"), "°C")
print("72 °F to Kelvin:  ", convert_temp(72, "F", "K"), "K")
print("5.00 in to cm:    ", convert_english_metric(5.00, "in", "cm"), "cm")
print("1.25 atm to mmHg: ", convert_english_metric(1.25, "atm", "mmHg"), "mmHg")
````

````python
# Tool 2: Stoichiometry, Limiting Reactant, Excess Remaining & Percent Yield Solver

def solve_limiting_reactant(r1_name, r1_g, r1_mm, r1_coeff,
                            r2_name, r2_g, r2_mm, r2_coeff,
                            prod_name, prod_mm, prod_coeff,
                            actual_yield_g=None):
    """
    Solves stoichiometry reaction problems:
    - Identifies Limiting and Excess reactants
    - Calculates Theoretical Yield of product (in grams and moles)
    - Calculates mass of Excess Reactant remaining unreacted
    - Calculates Percent Yield if actual experimental yield is provided
    """
    moles_r1 = r1_g / r1_mm
    moles_r2 = r2_g / r2_mm
    
    equiv_r1 = moles_r1 / r1_coeff
    equiv_r2 = moles_r2 / r2_coeff
    
    if equiv_r1 < equiv_r2:
        lr_name, xs_name = r1_name, r2_name
        theo_moles_prod = equiv_r1 * prod_coeff
        moles_xs_used = equiv_r1 * r2_coeff
        xs_rem_g = (moles_r2 - moles_xs_used) * r2_mm
    else:
        lr_name, xs_name = r2_name, r1_name
        theo_moles_prod = equiv_r2 * prod_coeff
        moles_xs_used = equiv_r2 * r1_coeff
        xs_rem_g = (moles_r1 - moles_xs_used) * r1_mm

    theo_yield_g = theo_moles_prod * prod_mm

    res = {
        "Limiting Reactant": lr_name,
        "Excess Reactant": xs_name,
        f"Theoretical Yield of {prod_name} (g)": round(theo_yield_g, 3),
        f"Theoretical Yield of {prod_name} (mol)": round(theo_moles_prod, 4),
        f"Excess {xs_name} Remaining (g)": round(xs_rem_g, 3)
    }
    if actual_yield_g is not None:
        pct = (actual_yield_g / theo_yield_g) * 100.0
        res["Actual Yield (g)"] = actual_yield_g
        res["Percent Yield (%)"] = f"{pct:.2f}%"
    return res

print("--- Tool 2: Stoichiometry & Limiting Reactant ---")
# N2 + 3 H2 -> 2 NH3 with 1.00 g N2 and 3.00 g H2, actual yield = 1.05 g NH3
result = solve_limiting_reactant(
    r1_name="N2", r1_g=1.00, r1_mm=28.02, r1_coeff=1,
    r2_name="H2", r2_g=3.00, r2_mm=2.016, r2_coeff=3,
    prod_name="NH3", prod_mm=17.03, prod_coeff=2,
    actual_yield_g=1.05
)
for k, v in result.items():
    print(f"  {k}: {v}")
````

````python
# Tool 3: Empirical and Molecular Formula Determiner

ATOMIC_MASSES = {
    "H": 1.008, "He": 4.003, "Li": 6.941, "Be": 9.012, "B": 10.811,
    "C": 12.011, "N": 14.007, "O": 15.999, "F": 18.998, "Ne": 20.180,
    "Na": 22.990, "Mg": 24.305, "Al": 26.982, "Si": 28.086, "P": 30.974,
    "S": 32.065, "Cl": 35.453, "Ar": 39.948, "K": 39.098, "Ca": 40.078,
    "Cr": 51.996, "Mn": 54.938, "Fe": 55.845, "Cu": 63.546, "Zn": 65.38,
    "Br": 79.904, "Ag": 107.868, "I": 126.904, "Ba": 137.327, "Pb": 207.2
}

def solve_empirical_formula(pct_dict, molar_mass=None):
    """
    Determines empirical and molecular formulas from mass or percentage composition.
    Recognizes common fractions (.5, .33, .67, .25, .75) to find whole-number formulas.
    """
    moles = {elem: pct / ATOMIC_MASSES[elem] for elem, pct in pct_dict.items()}
    min_m = min(moles.values())
    raw_ratios = {elem: m / min_m for elem, m in moles.items()}

    # Determine integer multiplier for fractional ratios
    multiplier = 1
    for r in raw_ratios.values():
        rem = r % 1
        if 0.15 < rem < 0.85:
            if abs(rem - 0.5) < 0.1:
                multiplier = max(multiplier, 2)
            elif abs(rem - 0.33) < 0.1 or abs(rem - 0.67) < 0.1:
                multiplier = max(multiplier, 3)
            elif abs(rem - 0.25) < 0.1 or abs(rem - 0.75) < 0.1:
                multiplier = max(multiplier, 4)

    final_subscripts = {elem: int(round(r * multiplier)) for elem, r in raw_ratios.items()}
    emp_formula = "".join([f"{e}{final_subscripts[e] if final_subscripts[e] > 1 else ''}" for e in final_subscripts])
    emp_mass = sum(final_subscripts[e] * ATOMIC_MASSES[e] for e in final_subscripts)

    print("Calculated Mole Ratios:", {e: round(r, 2) for e, r in raw_ratios.items()})
    print(f"Empirical Formula:       {emp_formula}")
    print(f"Empirical Formula Mass:  {emp_mass:.2f} g/mol")

    if molar_mass:
        n = int(round(molar_mass / emp_mass))
        mol_subscripts = {e: final_subscripts[e] * n for e in final_subscripts}
        mol_formula = "".join([f"{e}{mol_subscripts[e] if mol_subscripts[e] > 1 else ''}" for e in mol_subscripts])
        print(f"Molecular Multiplier n:  {n}")
        print(f"Molecular Formula:       {mol_formula} (Molar Mass: {molar_mass:.2f} g/mol)")

print("--- Tool 3: Empirical Formula Examples ---")
print("1. Cryolite (Practice Exam 1-A Question 12):")
solve_empirical_formula({"Na": 32.79, "Al": 13.02, "F": 54.19})

print("\n2. Glucose (40.0% C, 6.71% H, 53.29% O, MM = 180.16 g/mol):")
solve_empirical_formula({"C": 40.0, "H": 6.71, "O": 53.29}, molar_mass=180.16)
````

````python
# Tool 4: Electromagnetic Radiation Solver (c = lambda * nu, E = h * nu)

SPEED_OF_LIGHT = 2.9979e8   # m/s
PLANCK_CONST = 6.626e-34    # J*s
AVOGADRO_CONST = 6.022e23   # photons/mol

def solve_radiation(wavelength_m=None, frequency_hz=None, energy_j=None):
    """
    Solves radiation parameters from any one given input (wavelength, frequency, or photon energy).
    Returns wavelength (m, nm), frequency (Hz), photon energy (J), and molar energy (kJ/mol).
    """
    c, h, N_A = SPEED_OF_LIGHT, PLANCK_CONST, AVOGADRO_CONST
    
    if wavelength_m is not None:
        lam = wavelength_m
        nu = c / lam
        e = h * nu
    elif frequency_hz is not None:
        nu = frequency_hz
        lam = c / nu
        e = h * nu
    elif energy_j is not None:
        e = energy_j
        nu = e / h
        lam = c / nu
    else:
        return "Please provide at least one parameter."

    e_kj_mol = (e * N_A) / 1000.0
    lam_nm = lam * 1e9

    if lam < 1e-11: region = "Gamma Ray"
    elif lam < 1e-8: region = "X-Ray"
    elif lam < 4e-7: region = "Ultraviolet (UV)"
    elif lam <= 7.5e-7: region = "Visible Spectrum"
    elif lam < 1e-3: region = "Infrared (IR)"
    elif lam < 1.0: region = "Microwave"
    else: region = "Radio Waves"

    return {
        "Wavelength (m)": f"{lam:.3e} m",
        "Wavelength (nm)": f"{lam_nm:.2f} nm",
        "Frequency (Hz)": f"{nu:.3e} s⁻¹ (Hz)",
        "Energy per photon (J)": f"{e:.3e} J",
        "Energy per mole (kJ/mol)": f"{e_kj_mol:.2f} kJ/mol",
        "Spectral Region": region
    }

print("--- Tool 4: Electromagnetic Radiation Solver ---")
# Quiz Question 2 (wavelength 3.12e-3 cm = 3.12e-5 m)
res = solve_radiation(wavelength_m=3.12e-5)
for k, v in res.items():
    print(f"  {k}: {v}")
````

````python
# Tool 5: Bohr Hydrogen Transition Calculator
# Delta E = -R_H * (1/n_f^2 - 1/n_i^2)

RYDBERG_ENERGY = 2.178e-18  # J
SPEED_OF_LIGHT = 2.9979e8   # m/s
PLANCK_CONST = 6.626e-34    # J*s

def solve_bohr(n_i, n_f):
    """
    Calculates Delta E, photon wavelength (nm), transition type (absorbed vs emitted),
    and identifies the hydrogen spectral series (Lyman, Balmer, Paschen, Brackett).
    """
    R_H = RYDBERG_ENERGY
    c = SPEED_OF_LIGHT
    h = PLANCK_CONST

    delta_E = -R_H * (1.0 / (n_f**2) - 1.0 / (n_i**2))
    e_photon = abs(delta_E)
    nu = e_photon / h
    lam = c / nu
    lam_nm = lam * 1e9

    lower_n = min(n_i, n_f)
    series_map = {
        1: "Lyman Series (Ultraviolet)",
        2: "Balmer Series (Visible Spectrum)",
        3: "Paschen Series (Infrared)",
        4: "Brackett Series (Infrared)"
    }
    series = series_map.get(lower_n, "Higher Infrared Series")

    return {
        "Transition": f"n = {n_i} -> n = {n_f}",
        "Process": "Photon Emitted (Energy Released)" if delta_E < 0 else "Photon Absorbed (Energy Required)",
        "Delta E (J)": f"{delta_E:.3e} J",
        "Photon Energy (J)": f"{e_photon:.3e} J",
        "Wavelength (nm)": f"{lam_nm:.1f} nm",
        "Spectral Series": series
    }

print("--- Tool 5: Bohr Hydrogen Transitions ---")
print("1. Balmer Emission (n = 4 -> n = 2):")
for k, v in solve_bohr(4, 2).items():
    print(f"   {k}: {v}")

print("\n2. Ground State Excitation (n = 1 -> n = 5):")
for k, v in solve_bohr(1, 5).items():
    print(f"   {k}: {v}")
````

````python
# Tool 6: de Broglie Matter Wavelength Solver (lambda = h / (m * v))

PLANCK_CONST = 6.626e-34  # J*s

def de_broglie(mass_kg, velocity_ms):
    """
    Computes matter wavelength for a particle of mass (kg) moving at velocity (m/s).
    Returns wavelength in meters, nanometers, and picometers.
    """
    h = PLANCK_CONST
    lam = h / (mass_kg * velocity_ms)
    return {
        "Wavelength (m)": f"{lam:.3e} m",
        "Wavelength (nm)": f"{lam * 1e9:.3e} nm",
        "Wavelength (pm)": f"{lam * 1e12:.3f} pm"
    }

print("--- Tool 6: de Broglie Matter Wavelengths ---")
print("Alpha particle (6.64e-27 kg at 3.00e6 m/s):")
print(" ", de_broglie(6.64e-27, 3.00e6))

print("Electron in 1st Bohr Orbit (9.109e-31 kg at 2.20e6 m/s):")
print(" ", de_broglie(9.109e-31, 2.20e6))
````

````python
# Tool 7: Formal Charge, Bond Polarity & MO Bond Order Evaluator

def formal_charge(valence_e, lone_pair_e, bonds):
    """
    Formal Charge = Valence e⁻ - Nonbonding e⁻ (lone pair electrons) - Number of bonds
    """
    return valence_e - lone_pair_e - bonds

def bond_polarity(en1, en2):
    """
    Evaluates bond polarity based on Pauling Electronegativity Difference (Delta EN):
    - Delta EN <= 0.4: Nonpolar Covalent
    - 0.5 <= Delta EN < 2.0: Polar Covalent
    - Delta EN >= 2.0: Ionic
    """
    delta_en = abs(en1 - en2)
    if delta_en <= 0.4:
        b_type = "Nonpolar Covalent (equal sharing)"
    elif delta_en < 2.0:
        b_type = "Polar Covalent (unequal sharing, dipole moment exists)"
    else:
        b_type = "Ionic (complete electron transfer)"
    return {"Delta EN": round(delta_en, 2), "Bond Classification": b_type}

def bond_order_mo(bonding_e, antibonding_e, unpaired_e=0):
    """
    Bond Order = (Bonding e⁻ - Antibonding e⁻) / 2
    Magnetism: Paramagnetic if unpaired e⁻ > 0, else Diamagnetic
    """
    bo = (bonding_e - antibonding_e) / 2.0
    stability = "Stable bond" if bo > 0 else "Unstable / Does not exist"
    mag = "Paramagnetic (has unpaired e⁻; attracted to magnetic field)" if unpaired_e > 0 else "Diamagnetic (all e⁻ paired; repelled by magnet)"
    return {
        "Bond Order": bo,
        "Stability": stability,
        "Magnetic Property": mag
    }

print("--- Tool 7: Formal Charge, Polarity & MO Evaluator ---")
print("Formal charge on P in PO4³⁻ (Valence=5, Lone=0, Bonds=5):", formal_charge(5, 0, 5))
print("Formal charge on O with single bond in PO4³⁻ (Valence=6, Lone=6, Bonds=1):", formal_charge(6, 6, 1))

print("\nBond Polarity:")
print("  C-H bond (EN C=2.5, H=2.1):", bond_polarity(2.5, 2.1))
print("  H-F bond (EN H=2.1, F=4.0):", bond_polarity(2.1, 4.0))
print("  Na-Cl bond (EN Na=0.9, Cl=3.0):", bond_polarity(0.9, 3.0))

print("\nMolecular Orbital Theory (MOT):")
print("  O2 Molecule (8 bonding, 4 antibonding, 2 unpaired):", bond_order_mo(8, 4, unpaired_e=2))
print("  N2 Molecule (8 bonding, 2 antibonding, 0 unpaired):", bond_order_mo(8, 2, unpaired_e=0))
print("  He2 Molecule (2 bonding, 2 antibonding, 0 unpaired):", bond_order_mo(2, 2, unpaired_e=0))
````
